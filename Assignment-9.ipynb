```python
# Trigram Language Model for Text Generation

from collections import defaultdict, Counter
import random
import string

# -----------------------------
# STEP 1: LOAD & PREPROCESS TEXT
# -----------------------------
text = """I love deep learning and natural language processing.
Deep learning models can generate amazing text.
Language models learn patterns in text data."""

text = text.lower()
text = text.translate(str.maketrans('', '', string.punctuation))
tokens = text.split()

# -----------------------------
# STEP 2: BUILD A TRIGRAM MODEL
# -----------------------------
model = defaultdict(Counter)

for index in range(len(tokens) - 3):
    context = tuple(tokens[index:index + 3])
    next_word = tokens[index + 3]
    model[context][next_word] += 1

# -----------------------------
# STEP 3: GENERATE TEXT
# -----------------------------
def generate_text(seed_text, next_words=10):
    seed_tokens = seed_text.lower().translate(str.maketrans('', '', string.punctuation)).split()
    if len(seed_tokens) < 3:
        raise ValueError('Seed text must contain at least 3 words.')

    generated = seed_tokens[:]
    context = tuple(seed_tokens[-3:])

    for _ in range(next_words):
        choices = model.get(context)
        if not choices:
            break

        words, weights = zip(*choices.items())
        next_word = random.choices(words, weights=weights, k=1)[0]
        generated.append(next_word)
        context = tuple(generated[-3:])

    return ' '.join(generated)

# -----------------------------
# TEST
# -----------------------------
print(generate_text('i love deep', 5))
```
