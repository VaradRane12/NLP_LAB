```python

import re
from collections import defaultdict, Counter
```

```python

def preprocess(text):
    text = text.lower()
    text = re.sub(r'[^a-zA-Z\s]', '', text)  # remove punctuation
    tokens = text.split()
    return tokens
```

```python

def generate_ngrams(tokens, n):
    ngrams = []
    for i in range(len(tokens) - n + 1):
        ngram = tuple(tokens[i:i+n])
        ngrams.append(ngram)
    return ngrams
```

```python

def build_ngram_model(tokens, n):
    model = defaultdict(Counter)
    
    for i in range(len(tokens) - n + 1):
        context = tuple(tokens[i:i+n-1])
        next_word = tokens[i+n-1]
        model[context][next_word] += 1
    
    return model
```

```python

def calculate_probabilities(model):
    prob_model = {}
    
    for context, counter in model.items():
        total = sum(counter.values())
        prob_model[context] = {}
        
        for word, count in counter.items():
            prob_model[context][word] = count / total
    
    return prob_model
```

```python

def predict_next(prob_model, context):
    context = tuple(context)
    
    if context in prob_model:
        return max(prob_model[context], key=prob_model[context].get)
    else:
        return "No prediction available"
```

```python
    
text = "I love machine learning and I love deep learning"

# Preprocess
tokens = preprocess(text)
print("Tokens:", tokens)

# Generate N-grams
print("\nBigrams:", generate_ngrams(tokens, 2))
print("\nTrigrams:", generate_ngrams(tokens, 3))

# Build Bigram Model
bigram_model = build_ngram_model(tokens, 2)
bigram_prob = calculate_probabilities(bigram_model)

# Build Trigram Model
trigram_model = build_ngram_model(tokens, 3)
trigram_prob = calculate_probabilities(trigram_model)

# Prediction
print("\nBigram Prediction for 'i':",
      predict_next(bigram_prob, ['i']))

print("Bigram Prediction for 'love':",
      predict_next(bigram_prob, ['love']))

print("\nTrigram Prediction for 'i love':",
      predict_next(trigram_prob, ['i', 'love']))

print("Trigram Prediction for 'love deep':",
      predict_next(trigram_prob, ['love', 'deep']))
```
